```python
import requests
import os
import getpass
import json
```

```python
base_url = "https://api.whylabsapp.com"
org_id = "org-0"
dataset_id = "model-0"
```

```python
api_key = getpass.getpass("Enter API Key:")
```

```python
headers = {"Accept": "application/json", "Content-Type": "application/json", "X-API-KEY": api_key}
```

Get up to 3 anomalies from the 19th of March, in descending order.

```python
resp = requests.post(
    url=base_url + f"/v0/organizations/{org_id}/dataset/{dataset_id}/data/analysis-results",
    data=json.dumps({
      "interval": "2024-03-19T00:00:00Z/P1D",
      "onlyAnomalies": True,
      "offset": 0,
      "limit": 3,
      "order": "DESC"
    }),
    headers=headers
)
resp.json()
```

Get all drift analysis results for a specific column

```python
resp = requests.post(
    url=base_url + f"/v0/organizations/{org_id}/dataset/{dataset_id}/data/analysis-results",
    data=json.dumps({
      "interval": "2024-03-19T00:00:00Z/P5D",
      "columns": ["bc_util"],
      "analyzerTypes": ["drift"],
      "onlyAnomalies": False,
      "userMarkedUnhelpful": True,
      "offset": 0,
      "limit": 3,
      "order": "DESC"
    }),
    headers=headers
)
resp.json()
```

Get all results for a specific monitor, including failed analysis attempts and those marked as unhelpful

```python
resp = requests.post(
    url=base_url + f"/v0/organizations/{org_id}/dataset/{dataset_id}/data/analysis-results",
    data=json.dumps({
      "interval": "2024-03-19T00:00:00Z/P1D",
      "monitorIds": ["frequent-items-drift-monitor-bx6m80"],
      "onlyAnomalies": False,
      "userMarkedUnhelpful": True,
      "includeFailures": False,
      "offset": 0,
      "limit": 3,
      "order": "DESC"
    }),
    headers=headers
)
resp.json()
```

Use the whylabs client to make an analysis results request.

```python
# !pip install "whylabs-client~=0.6.0"
```

```python
import whylabs_client
from whylabs_client.api import data_api
from whylabs_client.model.analysis_results_request import AnalysisResultsRequest
from whylabs_client.model.sort_order import SortOrder
configuration = whylabs_client.Configuration(
    host = base_url
)
configuration.api_key['ApiKeyAuth'] = api_key

data_api = data_api.DataApi(whylabs_client.ApiClient(configuration))
```

```python
results = data_api.analysis_results_data(org_id, dataset_id, AnalysisResultsRequest(
      interval="2024-03-19T00:00:00Z/P1D",
      only_anomalies=True,
      include_failures=False,
      include_unhelpful=False,
      offset=0,
      limit=3,
      order=SortOrder("DESC")
))
```
